Importamos la librería de python para conectarnos a Neo4j

```python
# Importar la librería de neo4j
from neo4j import GraphDatabase
```

Abrimos una conexión a Neo4j a la instancia activa y la base de datos que querremos usar

```python
#Conexión a la base de datos a utilizar 
uri = "neo4j://127.0.0.1:7687"
user = "neo4j"
password = "12345678"
DB = "shv"

try:
    driver = GraphDatabase.driver(uri, auth=(user, password))

    with driver.session() as sesion:
        result = sesion.run("RETURN 1")
        if result.single()[0] ==1:
            print("Conexión exitosa a Neo4j")

except:
    print("Falló la conexión")
```

Creamos una consulta en cypher que use la coneixón activa y nos regrese el nombre de los Villanos como una lista

```python
query = """
    MATCH (v:Villian)
    RETURN v.Name AS Villanos;
"""

def run_read(tx, q, p=None):
    return list(tx.run(q, p or {}))

try:
    with driver.session(database=DB) as sesion:
        records = sesion.execute_read(run_read, query)
        rows = [r["Villanos"] for r in records]
        print(f"Villanos: \n {rows}")
       
finally:
    #driver.close()
    print("Conexión a Neo4j cerrada")        

```

Creamos una consulta en cypher que use la coneixón activa y nos regrese el nombre de los Villanos en una columna. Solo mostramos los primeros tres.

```python
query = """
    MATCH (v:Villian)
    RETURN v.Name AS Villanos
    LIMIT 3;
"""

with driver.session(database=DB) as sesion:
    records = sesion.execute_read(run_read, query)
    rows = [r["Villanos"] for r in records]

print("Villanos")
print("--------")
for Villanos in rows:
    print(Villanos)
```

#### Creamos una conexión a Neo4j usando el driver Graph Data Science

Importamos las librerías necearias

```python
import pandas as pd

from graphdatascience import GraphDataScience
```

Creamos una conexión de python a Neo4j usando la librería Graph Data Science

```python
uri = "neo4j://127.0.0.1:7687"
user ="neo4j"
password = "12345678"
DB = "shv"

try:
    gds = GraphDataScience(uri, auth=(user, password))

    gds.set_database(DB)

    test = gds.run_cypher("RETURN 1 as prueba")
    if test['prueba'].iloc[0] ==1:
        print("Conexión exitosa a Neo4j")

except:
    print("Falló la conexión")
```

Corremos una consulta simple en cypher y mostramos el resultado

```python
gds.run_cypher("RETURN 'Hola Mundo' AS Mensaje")
```

Corremos la consulta en cypher que muestra los villanos en una columna. Mostramos solo los primeros tres.

```python
query = """
    MATCH (v:Villian)
    RETURN v.Name AS Villanos
    LIMIT 3;
"""

gds.run_cypher(query)
```

Creamos una consulta que muestre el total de Villanos

```python
query = """
    MATCH (v:Villian)
    RETURN COUNT(DISTINCT v.Name) AS NumVillanos;
"""

gds.run_cypher(query)
```
