```python
import pandas as pd
import os
```

```python
dir = './data/'

csvsNodes = []
csvsEdges = []
```

```python
# Leer y clasificar archivos
for i in os.listdir(dir):
    path = os.path.join(dir, i)
    
    if path.endswith('.csv'):
        df = pd.read_csv(path)
        
        if "nodes" in path.lower():
            csvsNodes.append(df)
        else:
            csvsEdges.append(df)
```

```python
# Fusionar y guardar todos los nodes
if csvsNodes:
    df_nodes_combined = pd.concat(csvsNodes, ignore_index=True)

    df_nodes_combined = df_nodes_combined.drop_duplicates(subset=['Id'])
    df_nodes_combined.to_csv('./nodes_fusionados.csv', index=False)
    print(f"Nodes fusionados guardados: {len(df_nodes_combined)} nodos únicos")
    print(f"Archivo: nodes_fusionados.csv")
```

```python
# Fusionar y guardar todos los edges
if csvsEdges:
    df_edges_combined = pd.concat(csvsEdges, ignore_index=True)

    # df_edges_combined = df_edges_combined.drop_duplicates()
    df_edges_combined.to_csv('./edges_fusionados.csv', index=False)
    print(f"Edges fusionados guardados: {len(df_edges_combined)} relaciones")
    print(f"Archivo: edges_fusionados.csv")
```
